```python
def parse_matches(value):
    if pd.isna(value) or str(value).strip() == "":
        return set()
    
    return set(
        x.strip()
        for x in str(value).split(",")
        if x.strip()
    )

gt["true_matches"] = gt["matched_entity_ids"].apply(parse_matches)
```

```python
gt[["source1_entity_id", "true_matches"]].head()
```

```python
def calculate_f05(true_matches, predicted_matches):
    true_matches = set(true_matches)
    predicted_matches = set(predicted_matches)

    tp = len(true_matches & predicted_matches)
    fp = len(predicted_matches - true_matches)
    fn = len(true_matches - predicted_matches)

    precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0.0

    beta = 0.5

    if precision == 0 and recall == 0:
        f05 = 0.0
    else:
        f05 = (
            (1 + beta**2) * precision * recall
            / ((beta**2 * precision) + recall)
        )

    return precision, recall, f05
```

```python
calculate_f05(
    {"S2-A", "S2-B"},
    {"S2-A", "S2-C"}
)
```

```python
print("Exact:", calculate_f05({"A", "B"}, {"A", "B"}))

print("Wrong:", calculate_f05({"A", "B"}, {"C", "D"}))

print("Partial:", calculate_f05({"A", "B"}, {"A", "C"}))

print("True no-match:", calculate_f05(set(), set()))

print("False positive on no-match:", calculate_f05(set(), {"A"}))
```

```python
# Take 5 real S1 entities
test_gt = gt[["source1_entity_id", "true_matches"]].head(5).copy()

# Create deliberately imperfect predictions
test_predictions = {
    test_gt.iloc[0]["source1_entity_id"]: test_gt.iloc[0]["true_matches"],  # perfect
    test_gt.iloc[1]["source1_entity_id"]: set(),                           # predict no match
    test_gt.iloc[2]["source1_entity_id"]: {"FAKE-ID"},                     # wrong
    test_gt.iloc[3]["source1_entity_id"]: set(list(test_gt.iloc[3]["true_matches"])[:1]), # partial
    test_gt.iloc[4]["source1_entity_id"]: test_gt.iloc[4]["true_matches"] | {"FAKE-ID"}   # extra false positive
}
```

```python
results = []

for _, row in test_gt.iterrows():
    s1_id = row["source1_entity_id"]
    true_matches = row["true_matches"]
    predicted_matches = test_predictions.get(s1_id, set())

    precision, recall, f05 = calculate_f05(
        true_matches,
        predicted_matches
    )

    results.append({
        "source1_entity_id": s1_id,
        "precision": precision,
        "recall": recall,
        "f05": f05
    })

results_df = pd.DataFrame(results)

results_df
```

```python
def evaluate_predictions(ground_truth, predictions):
    """
    Evaluate predictions using per-S1 precision, recall and F0.5,
    then macro-average across all S1 entities.

    predictions:
        Dictionary:
        {
            "S1-ID": {"S2-ID", "S3-ID", ...}
        }
    """

    results = []

    for _, row in ground_truth.iterrows():

        s1_id = row["source1_entity_id"]
        true_matches = row["true_matches"]

        # If our model didn't predict anything for this S1,
        # treat it as an empty prediction.
        predicted_matches = predictions.get(s1_id, set())

        precision, recall, f05 = calculate_f05(
            true_matches,
            predicted_matches
        )

        results.append({
            "source1_entity_id": s1_id,
            "precision": precision,
            "recall": recall,
            "f05": f05
        })

    results_df = pd.DataFrame(results)

    metrics = {
        "macro_precision": results_df["precision"].mean(),
        "macro_recall": results_df["recall"].mean(),
        "macro_f05": results_df["f05"].mean()
    }

    return results_df, metrics
```

```python
results_df, metrics = evaluate_predictions(
    test_gt,
    test_predictions
)

print("Macro Precision:", metrics["macro_precision"])
print("Macro Recall:", metrics["macro_recall"])
print("Macro F0.5:", metrics["macro_f05"])
```

```python
def validate_submission(predictions, test_s1, test_s2, test_s3):

    print("========== SUBMISSION CHECK ==========")

    # Required columns
    required = {
        "source1_entity_id",
        "matched_entity_ids"
    }

    missing_columns = required - set(predictions.columns)

    if missing_columns:
        print("❌ Missing columns:", missing_columns)
        return False

    print("✅ Required columns present")

    # S1 IDs
    required_s1 = set(test_s1["entity_id"])
    predicted_s1 = set(predictions["source1_entity_id"])

    # Duplicates
    duplicates = predictions[
        "source1_entity_id"
    ].duplicated().sum()

    print("Duplicate S1 rows:", duplicates)

    if duplicates > 0:
        print("❌ Duplicate S1 IDs")
        return False

    # Missing S1
    missing_s1 = required_s1 - predicted_s1

    print("Missing S1 IDs:", len(missing_s1))

    if missing_s1:
        print("❌ Missing test S1 entities")
        return False

    # Extra S1
    extra_s1 = predicted_s1 - required_s1

    print("Extra S1 IDs:", len(extra_s1))

    if extra_s1:
        print("❌ Unknown S1 IDs")
        return False

    # Valid S2/S3 IDs
    valid_targets = set(test_s2["entity_id"]) | set(test_s3["entity_id"])

    invalid_targets = set()

    for value in predictions["matched_entity_ids"].fillna(""):

        value = str(value).strip()

        if not value:
            continue

        for entity_id in value.split(","):

            entity_id = entity_id.strip()

            if entity_id and entity_id not in valid_targets:
                invalid_targets.add(entity_id)

    print("Invalid S2/S3 IDs:", len(invalid_targets))

    if invalid_targets:
        print("❌ Invalid target IDs")
        print(list(invalid_targets)[:10])
        return False

    print("\n✅ ALL BASIC CHECKS PASSED")

    return True
```

```python
def inspect_prediction_file(path):
    pred = pd.read_csv(path, sep="\t")

    print("========== RAJINI FILE ==========")
    print("Shape:", pred.shape)
    print("Columns:", pred.columns.tolist())
    print("\nFirst 5 rows:")
    display(pred.head())

    print("\nMissing values:")
    print(pred.isna().sum())

    print("\nDuplicate rows:")
    print(pred.duplicated().sum())

    if "source1_entity_id" in pred.columns:
        print(
            "\nUnique S1 IDs:",
            pred["source1_entity_id"].nunique()
        )

    return pred
```

```python
def prediction_summary(predictions):

    print("========== PREDICTION SUMMARY ==========")

    predictions = predictions.copy()

    predictions["matched_entity_ids"] = (
        predictions["matched_entity_ids"]
        .fillna("")
        .astype(str)
        .str.strip()
    )

    match_counts = predictions["matched_entity_ids"].apply(
        lambda x: 0 if not x else len(x.split(","))
    )

    print("Total S1 rows:", len(predictions))
    print("Unique S1 IDs:", predictions["source1_entity_id"].nunique())

    print("\nS1 entities with 0 matches:",
          (match_counts == 0).sum())

    print("S1 entities with >=1 match:",
          (match_counts >= 1).sum())

    print("\nTotal predicted matches:",
          match_counts.sum())

    print("\nMatch-count distribution:")
    print(match_counts.value_counts().sort_index())

    return match_counts
```

```python
def check_target_sources(predictions, test_s2, test_s3):

    valid_s2 = set(test_s2["entity_id"])
    valid_s3 = set(test_s3["entity_id"])

    s2_count = 0
    s3_count = 0
    invalid_count = 0

    for value in predictions["matched_entity_ids"].fillna(""):

        value = str(value).strip()

        if not value:
            continue

        for entity_id in value.split(","):

            entity_id = entity_id.strip()

            if entity_id.startswith("S2-"):
                s2_count += 1

            elif entity_id.startswith("S3-"):
                s3_count += 1

            else:
                invalid_count += 1

    print("S2 predictions:", s2_count)
    print("S3 predictions:", s3_count)
    print("Invalid-format predictions:", invalid_count)
```
